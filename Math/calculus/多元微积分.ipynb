## 多元微积分

```python
import matplotlib.pyplot as plt
import numpy as np
```

多元函数

一个自变量x与一个因变量f(x)相关联我们称为一元函数,比如的$f(x)=x^2$.多元函数则是多个自变量与一个因变量f(x,y)的关系[这里用二元方便可视化],比如$f(x,y)=x^2+y^2$

```python
fig=plt.figure(figsize=(8,4))
ax1=fig.add_subplot(1,2,1);
ax2=fig.add_subplot(1,2,2,projection='3d')
x=np.linspace(-2,2)
ax1.plot(x,x**2)
xx,yy=np.meshgrid(x,x)
ax2.plot_surface(xx,yy,xx**2+yy**2,cmap='viridis',alpha=0.8);
```

有上面的图形可见,一元函数有xy两个变量用2d图像表示,x-表示自变量,y-表示因变量,绘制出来的线条表示他们之间的关系.二元函数涉及3个变量需要3d图片表示,两个自变量分别对应两条坐标轴,它们张成一个x-y的定义域平面,因变量z则是在平面上方或下方高度.如果用水平平面 z = c 去切割三维曲面,切割得到的交线垂直投影到 x-y 平面上，所得到的曲线（或曲线集）称为函数 f(x, y) 高度为 c 的等高线。这条线上的所有点，其函数值（高度）都等于同一个常数 c。

```python
plt.contourf(xx,yy,xx**2+yy**2)
ax=plt.gca()
ax.set_aspect('equal')
```

上面的圆就是原函数$f(x,y)=x^2+y^2$三维曲面的等高线,等高线是一种降维可视化方法,通过固定一个变量减少维度,当固定z=c原来的函数就变成$x^2+y^2=c$就只有两个变量,也变成圆的表达式.

偏微分

f是x,y两个变量的函数,如果一个y是常数,y=$y_0$那么f(x,$y_0$)就是一个只有单个变量x的函数,它在x=$x_0$处的导数称为f关于x在($x_0,y_0$)处的偏导数,记$f_x(x_0,y_0)=\lim_{\Delta x \to \infty }\frac{f(x_0+\Delta x,y_0)-f(x_0,y_0)}{\Delta x}$,同理在y的偏导数当x为常数x=$x_0$则在y=$y_0$的偏导数$f_y(x_0,y_0)=\lim_{\Delta x \to \infty }\frac{f(x_0 ,y_0+\Delta y)-f(x_0,y_0)}{\Delta y}$

```python
fig,ax=plt.subplots(subplot_kw={'projection':'3d'})
zz=xx**2+yy**2

x0,y0=1,1
dx=2*x0
dy=2*y0
z0=x0**2+y0**2
# grad_x z-z0=grad_x(x-x0)
line_x=np.linspace(0,2,20)
line_y_const=np.ones_like(line_x)*y0
line_z_x=dx*(line_x-x0)+z0
# 曲面 x
curve_x=np.linspace(-1,3,20)
curve_y_const=np.ones_like(curve_x)*y0
curve_z_x=curve_x**2+curve_y_const**2


# grad_y z-z0=grad_y(y-y0)
line_y=np.linspace(0,2,20)
line_x_const=np.ones_like(line_y)*x0
line_z_y=dy*(line_y-y0)+z0
# 曲面y
curve_y=np.linspace(-1,3,20)
curve_x_const=np.ones_like(curve_y)*x0
curve_z_y=curve_y**2+curve_x_const**2



ax.plot_surface(xx,yy,zz,alpha=0.5,cmap='viridis',edgecolor=None)
ax.scatter([x0],[y0],[z0],color='red',s=10,label='(1,1,2)')

ax.plot(curve_x,curve_y_const,curve_z_x,color='blue',linewidth=1,linestyle='--',)
ax.plot(curve_x_const,curve_y,curve_z_y,color='green',linewidth=1,linestyle='--',)


ax.plot(line_x,line_y_const,line_z_x,color='blue',linewidth=1)
ax.plot(line_x_const,line_y,line_z_y,color='green',linewidth=1)

ax.view_init(elev=25, azim=35)
```

极限与连续

$$
\lim_{(x,y)->(a,b)}f(x,y)=L
$$
二元函数的极限有不同含义,f(x,y)在(x,y)趋近于(a,b)时趋近L,对于一元函数都是在一x轴上易于区分,二元则不同上面讲了二元函数的定义域是一个平面.可以通过画圆的方式来找范围,1.在XY平面上以目标点(a,b)为圆心画个小圆,这个半径为$\delta$,圆的作用限制(x,y)的活动范围.2.只要(x,y)落在这个小圆内($\sqrt{(x-a)^2-(y-b)^2<\delta>}$),那么曲面上对应的点f(x,y)的高度必须落在L附近即｜f(x,y)-L｜< ϵ

> 无论我要求你离 L 有多近（无论 ϵ 多小），我总能在 (a,b)点周围找到一个小圆盘（半径 δ），只要点 (x,y)踏进这个小圆盘（距离小于 δ，但不包括圆心本身），函数值就绝对满足我的要求。

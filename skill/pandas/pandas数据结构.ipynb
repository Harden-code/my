两种结构
- Series 一位数组+索引 (k-v形式)
- DataFrame 二位的数据表格+行索引+列索引
> DataFrame是Series的容器

axis 轴 0行方向 1列方向

索引:自然索引(0-1),自己创建索引

```python
pd.Series(data=['a','b']) # 自然索引
```

```python
pd.Series([1,2,3],index=['a','b','c']) # 指定索引
```

```python
pd.Series(2,index=['a','b']) # 会根据索引进行填充
```

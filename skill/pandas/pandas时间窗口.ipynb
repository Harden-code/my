时间窗口
不计算全局数据的总和或平均,而是计算局部移动区间内的数据,核心目的是平滑短期波动,凸显长期趋势

1. rolling 滑动窗口
    - winodw 窗口大小,可传数字3行数据,可传字符3D表3天(按天算中间日期缺失会自动识别)
    - min_periods 最小观测值,如果窗口为3前两行凑不够就会Nan
2. expanding 累计窗口 窗口不断扩大,起点第一行(计算记录最高销售)
3. ewn 指数加权窗口

```python
dates = pd.date_range('2023-01-01', periods=5)
df = pd.DataFrame({'销量': [10, 20, 30, 40, 50]}, index=dates)

df.rolling(2,min_periods=1).mean()
```

```python
'''
销量  历史均值  历史最高
2023-01-01  10  10.0   10.0  <-- (10)/1
2023-01-02  20  15.0   20.0  <-- (10+20)/2
2023-01-03  30  20.0   30.0  <-- (10+20+30)/3
2023-01-04  40  25.0   40.0  <-- (10+20+30+40)/4
2023-01-05  50  30.0   50.0  <-- 每天的计算范围都在扩大
'''
df.expanding().sum()
```

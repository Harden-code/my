```python
import pandas as pd
import numpy as np
```

创建

```python
# 直接创建
pd.Series(['a','b','c','c','b'],dtype='category')
```

```python
# 通过创建分类数据
pd.Categorical(['x','y','x','z'],categories=['x','y','z'])
```

```python
# 创建categoricalDtype方式
ctype=pd.api.types.CategoricalDtype(['x','y','z'])

pd.Series(['x','y','z','x','x'],dtype=ctype)
```

```python
# 直接通过类型转换
s=pd.Series(['x','y','z','x','x'])
s=s.astype('category')
```

```python
# cut 数据分箱也会创建分类
x=pd.Series(np.random.randint(1,10,(100,)))
pd.cut(x,bins=[1,4,5,10])
```

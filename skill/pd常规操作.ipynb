```python
import pandas as pd
import numpy as np
import plotly.express as pe
```

读取数据操作
可以从csv,excel,json,db里读取数据,方法对应read_xxx

输出则对应的DataFrame里的to_xxx

```python
iris=pe.data.iris()
iris.to_csv('iris.csv')
```

读取常用参数
- usecols 读取哪些列
- names 指定列名
- header 头行
- sep 分隔符
- converters 列数据转换
- true_values=['yes','1'] /false_values=['no','2']
- skiprows 跳过几行
- skip_blanl_lines 跳过空行
- na_values 认为是空值的值

```python


df=pd.read_csv('iris.csv',header=0,usecols=['idx','sepal_length','species_id'],index_col=['idx'],converters={'species_id':lambda id:int(id)-1},
            #    skiprows=2,
               skip_blank_lines=True,na_values={'1.#f'})
df.head()
```

索引
set_index设置,可设置双层索引,设置索引后不会对修改原有的df,可通过设置inplace来生效

reset_index重制索引,也有inplace参数

常见索引
- RangeIndex 默认的整数索引
- Index 基础索引,普通字符串,整数
- MultiIndex 多层索引,处理高纬
- DatatimeIndex 事件序列索引
- CategoricalIndex 分类索引,自定义排序
> 索引为了方便数据的查找

```python
cate_index=pd.CategoricalIndex(['S','M','L'],ordered=True)

# size 是默认属性
s=pd.DataFrame(data={'num':[1,2,4,5,6],'sizes':['S','M','M','L','S']})
# 创建分类
s.sizes=pd.Categorical(s.sizes,categories=cate_index)
s.sizes.dtypes
```
